```python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Subir el archivo Excel
archivos = files.upload()
nombre_archivo = next(iter(archivos))

# Leer la hoja del catálogo
df = pd.read_excel(
    nombre_archivo,
    sheet_name="Catalogo combustibles"
)

# Limpiar los nombres de las columnas
df.columns = df.columns.str.strip()

print("Base de datos cargada correctamente")
print("Cantidad de combustibles:", len(df))
display(df.head())
```

```python

# GRÁFICO 1: ESTADO FÍSICO

columna = "Estado / tipo físico"
datos = df[columna].value_counts()

fig, ax = plt.subplots(figsize=(9, 5))

barras = ax.bar(datos.index, datos.values, width=0.6)

ax.set_title(
    "Distribución de los combustibles por estado físico",
    fontsize=15, fontweight="bold", pad=18
)
ax.set_xlabel("Estado físico")
ax.set_ylabel("Cantidad de combustibles")
ax.bar_label(barras, padding=4, fontsize=11)
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("grafico_1_estado_fisico.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

# GRÁFICO 2: SECTOR DE USO PRINCIPAL

def clasificar_sector(nombre):
    n = str(nombre).lower()

    if any(x in n for x in [
        "turbo", "jet a-1", "avgas", "aviación"
    ]):
        return "Aviación"

    if any(x in n for x in ["marino"]):
        return "Transporte marítimo"

    if any(x in n for x in [
        "glp automotor", "gnv", "gasolina", "gasohol",
        "diésel", "diesel", "biodiésel", "biodiesel"
    ]):
        return "Transporte terrestre"

    if any(x in n for x in [
        "carbón", "carbon", "fuel oil", "petróleo industrial",
        "petroleo industrial", "pellets", "hidrógeno", "hidrogeno"
    ]):
        return "Industria y energía"

    if any(x in n for x in [
        "glp envasado", "glp a granel", "queroseno", "leña", "lena"
    ]):
        return "Hogares y comercio"

    if any(x in n for x in ["biogás", "biogas", "gas natural"]):
        return "Hogares, comercio e industria"

    return "Otros usos"

df["Sector de uso"] = df["Nombre del combustible"].apply(
    clasificar_sector
)

datos = df["Sector de uso"].value_counts().sort_values()

fig, ax = plt.subplots(figsize=(10, 6))

barras = ax.barh(datos.index, datos.values, height=0.65)

ax.set_title(
    "Clasificación de combustibles por sector de uso",
    fontsize=15, fontweight="bold", pad=16
)
ax.set_xlabel("Cantidad de combustibles")
ax.set_ylabel("Sector de uso")
ax.bar_label(barras, padding=4, fontsize=10)
ax.grid(axis="x", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("grafico_2_sector_uso.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

# GRÁFICO 3: ORIGEN ENERGÉTICO

def clasificar_origen(nombre):
    n = str(nombre).lower()

    if any(x in n for x in [
        "biodiésel", "biodiesel", "etanol", "biogás",
        "biogas", "pellets de biomasa", "leña", "lena"
    ]):
        return "Biocombustibles y biomasa"

    if any(x in n for x in [
        "hidrógeno", "hidrogeno"
    ]):
        return "Combustibles alternativos"

    if any(x in n for x in [
        "gasolina", "gasohol", "diésel", "diesel",
        "glp", "gnv", "gas natural", "turbo",
        "jet a-1", "avgas", "queroseno",
        "fuel oil", "petróleo industrial",
        "petroleo industrial", "carbón", "carbon",
        "combustible marino"
    ]):
        return "Combustibles fósiles"

    return "Otros o clasificación por verificar"

df["Origen energético"] = df["Nombre del combustible"].apply(
    clasificar_origen
)

datos = df["Origen energético"].value_counts()

fig, ax = plt.subplots(figsize=(8, 7))

ax.pie(
    datos.values,
    labels=datos.index,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.72,
    wedgeprops={"width": 0.42, "edgecolor": "white"},
    textprops={"fontsize": 10}
)

ax.set_title(
    "Composición del catálogo por origen energético",
    fontsize=15, fontweight="bold", pad=20
)

plt.tight_layout()
plt.savefig("grafico_3_origen_energetico.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

# GRÁFICO 4: UNIDADES DE MEDIDA

def clasificar_unidad(unidad):
    u = str(unidad).lower()

    if "galón" in u or "galon" in u:
        return "Galón"
    elif "litro" in u:
        return "Litro"
    elif "kilogramo" in u or "kg" in u:
        return "Kilogramo"
    elif "metro cúbico" in u or "m³" in u:
        return "Metro cúbico"
    elif "tonelada" in u:
        return "Tonelada"
    elif "balón" in u or "balon" in u:
        return "Balón"
    else:
        return "Otras unidades"

df["Unidad agrupada"] = df["Unidad de medida habitual"].apply(
    clasificar_unidad
)

datos = df["Unidad agrupada"].value_counts().sort_values()

fig, ax = plt.subplots(figsize=(9, 5.5))

barras = ax.barh(datos.index, datos.values, height=0.65)

ax.set_title(
    "Unidades de medida utilizadas en los combustibles",
    fontsize=15, fontweight="bold", pad=18
)
ax.set_xlabel("Cantidad de registros")
ax.set_ylabel("Unidad de medida")
ax.bar_label(barras, padding=4, fontsize=10)
ax.grid(axis="x", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("grafico_4_unidades_medida.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from google.colab import files
import os

# Nombres de los cuatro gráficos creados anteriormente
graficos = [
    "grafico_1_estado_fisico.png",
    "grafico_2_sector_uso.png",
    "grafico_3_origen_energetico.png",
    "grafico_4_unidades_medida.png"
]

# Nombre del archivo PDF
nombre_pdf = "Reporte_Graficos_Tipos_de_Combustibles.pdf"

# Crear el PDF con una página por gráfico
with PdfPages(nombre_pdf) as pdf:
    for archivo in graficos:
        if not os.path.exists(archivo):
            raise FileNotFoundError(
                f"No se encuentra {archivo}. "
                "Primero ejecuta el código que genera los cuatro gráficos."
            )

        imagen = plt.imread(archivo)

        fig, ax = plt.subplots(figsize=(11.69, 8.27))
        ax.imshow(imagen)
        ax.axis("off")

        fig.suptitle(
            "ANÁLISIS DE LOS TIPOS DE COMBUSTIBLES EN EL PERÚ",
            fontsize=15,
            fontweight="bold",
            y=0.96
        )

        plt.tight_layout(rect=[0.02, 0.02, 0.98, 0.91])
        pdf.savefig(fig, bbox_inches="tight")
        plt.close(fig)

print("¡PDF creado correctamente!")

# Descargar automáticamente el PDF
files.download(nombre_pdf)
```
